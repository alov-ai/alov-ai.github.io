## 🧮 Elliptik əyrilərin riyazi əsasları (15 dəq)

<div style="background-color: #f0f8ff; padding: 15px; border-radius: 10px; border-left: 5px solid #4682b4;">
<h4>📐 Qısa Weierstrass tənliyi</h4>
<p>Xarakteristikası 2 və 3 olmayan meydanda qısa Weierstrass forması belədir:</p>
<p style="text-align: center; font-size: 1.2em;">$$E: y^2 = x^3 + ax + b.$$</p>
<p>Əyrinin qeyri-singulyar olması üçün $4a^3+27b^2\neq0$ olmalıdır. Standart diskriminant $\Delta=-16(4a^3+27b^2)$-dir; hər iki yazılış eyni qeyri-sıfır şərtini verir.</p>
<p><b>Vacib ayrım:</b> $E(\mathbb{R})$ həndəsi təsvir üçün sonsuz nöqtələr çoxluğudur; $E(\mathbb{F}_p)$ isə kriptoqrafiyada istifadə edilən sonlu qrupdur. Toplama düsturları oxşardır, amma $\mathbb{F}_p$-də bölmə modul tərsinə vurma deməkdir.</p>
</div>


````python
def weierstrass_discriminant(a, b):
    """char(F) != 2, 3 üçün qısa Weierstrass diskriminantı."""
    return -16 * (4 * a**3 + 27 * b**2)


examples = [
    ("y² = x³ - x", -1, 0, True),
    ("y² = x³ + 1", 0, 1, True),
    ("y² = x³ - 3x + 3", -3, 3, True),
    ("y² = x³", 0, 0, False),
]

for name, a, b, expected in examples:
    delta = weierstrass_discriminant(a, b)
    assert (delta != 0) is expected
    print(f"{name:24} Δ = {delta:5}; elliptik əyri: {expected}")
````

### 5.1 Elliptik əyrilərin növləri

* **Həqiqi əyrilər:** $a, b \in \mathbb{R}$ — qrafiki çəkmək üçün əlverişli
* **Sonlu meydan əyriləri:** $a, b \in \mathbb{F}_p$ ($p$ sadə ədəd) — kriptoqrafiyada istifadə olunur
* **İkili meydan əyriləri:** $a, b \in \mathbb{F}_{2^m}$ — bəzi aparat implementasiyalarında

### ✍️ Çalışma 1: Elliptik əyrilərin təyini (1 bal)

Aşağıdakı əyrilərin diskriminantlarını hesablayın və onların elliptik əyri olub-olmadığını müəyyən edin:

1. $y^2 = x^3 - x$ ($a=-1$, $b=0$)
2. $y^2 = x^3 + 1$ ($a=0$, $b=1$)
3. $y^2 = x^3 - 3x + 3$ ($a=-3$, $b=3$)
4. $y^2 = x^3$ ($a=0$, $b=0$) — bu elliptik əyridirmi?

````python
# Çalışma 1 — yoxlanılan cavablar
print("📝 ÇALIŞMA 1 CAVABLARI")
print("=" * 80)

for number, (name, a, b, expected) in enumerate(examples, 1):
    reduced_term = 4 * a**3 + 27 * b**2
    delta = weierstrass_discriminant(a, b)
    is_elliptic = delta != 0
    assert is_elliptic is expected
    print(f"{number}. {name}: 4a³+27b²={reduced_term}, Δ={delta}; "
          f"elliptik əyridir: {is_elliptic}")
````

## ➕ Nöqtə toplama əməliyyatı (25 dəq)

<div style="background-color: #f0f8ff; padding: 15px; border-radius: 10px; border-left: 5px solid #4682b4;">
<h4>📐 Nöqtə toplama (geometrik izah)</h4>
<p>Elliptik əyri üzərində $P$ və $Q$ nöqtələri verildikdə, $P + Q$ aşağıdakı kimi tapılır:</p>
<ol>
    <li>$P$ və $Q$-dən keçən düz xətt çəkilir</li>
    <li>Bu xətt əyrini üçüncü bir nöqtədə kəsir ($R$)</li>
    <li>$P + Q$, $R$-in $x$-oxuna nəzərən simmetrik nöqtəsidir ($-R$)</li>
</ol>
</div>

<div style="background-color: #e8f5e9; padding: 15px; border-radius: 10px; border-left: 5px solid #4caf50; margin-top: 10px;">
<h4>📌 Nümunə</h4>
<p>$y^2 = x^3 - x$ əyrisində $P = (-1, 0)$ və $Q = (0, 0)$ nöqtələrini toplayın.</p>
<p>Bu iki nöqtədən keçən xətt $y = 0$-dır. Bu xətt əyrini $x = -1, 0, 1$ nöqtələrində kəsir. Üçüncü kəsişmə nöqtəsi $R = (1, 0)$-dır. $R$-in $x$-oxuna nəzərən simmetriki yenə özü olduğu üçün $P + Q = (1, 0)$.</p>
</div>

<div style="background-color: #f0f8ff; padding: 15px; border-radius: 10px; border-left: 5px solid #4682b4; margin-top: 10px;">
<h4>📐 Nöqtə toplama (cəbri izah)</h4>
<p>$P = (x_1, y_1)$ və $Q = (x_2, y_2)$ nöqtələri üçün $P + Q = (x_3, y_3)$:</p>

<p><b>Əgər $P \neq Q$ olarsa:</b></p>
$$
\begin{align*}
\lambda &= \frac{y_2 - y_1}{x_2 - x_1} \\
x_3 &= \lambda^2 - x_1 - x_2 \\
y_3 &= \lambda(x_1 - x_3) - y_1
\end{align*}
$$

<p><b>Əgər $P = Q$ olarsa (nöqtəni ikiqat artırma):</b></p>
$$
\begin{align*}
\lambda &= \frac{3x_1^2 + a}{2y_1} \\
x_3 &= \lambda^2 - 2x_1 \\
y_3 &= \lambda(x_1 - x_3) - y_1
\end{align*}
$$
</div>

````python
class EllipticCurve:
    """Q üzərində y² = x³ + ax + b əyrisi; O nöqtəsi None ilə göstərilir."""

    def __init__(self, a, b):
        self.a = Fraction(a)
        self.b = Fraction(b)
        self.discriminant = weierstrass_discriminant(self.a, self.b)
        if self.discriminant == 0:
            raise ValueError("Diskriminant sıfırdır: əyri singulyardır")

    @staticmethod
    def _point(point):
        if point is None:
            return None
        if not isinstance(point, tuple) or len(point) != 2:
            raise TypeError("Nöqtə (x, y) cütü və ya None olmalıdır")
        return (Fraction(point[0]), Fraction(point[1]))

    def is_on_curve(self, x, y):
        x, y = Fraction(x), Fraction(y)
        return y**2 == x**3 + self.a * x + self.b

    def _validate(self, point):
        point = self._point(point)
        if point is not None and not self.is_on_curve(*point):
            raise ValueError(f"{point} nöqtəsi əyri üzərində deyil")
        return point

    def negate(self, point):
        point = self._validate(point)
        return None if point is None else (point[0], -point[1])

    def add(self, p1, p2):
        p1, p2 = self._validate(p1), self._validate(p2)
        if p1 is None:
            return p2
        if p2 is None:
            return p1

        x1, y1 = p1
        x2, y2 = p2
        if x1 == x2 and y1 == -y2:
            return None
        if p1 == p2:
            slope = (3 * x1**2 + self.a) / (2 * y1)
        else:
            slope = (y2 - y1) / (x2 - x1)

        x3 = slope**2 - x1 - x2
        result = (x3, slope * (x1 - x3) - y1)
        assert self.is_on_curve(*result)
        return result

    def multiply(self, point, scalar):
        point = self._validate(point)
        if isinstance(scalar, bool) or not isinstance(scalar, int):
            raise TypeError("Skalyar tam ədəd olmalıdır")
        if scalar < 0:
            return self.multiply(self.negate(point), -scalar)

        result, addend = None, point
        while scalar:
            if scalar & 1:
                result = self.add(result, addend)
            addend = self.add(addend, addend)
            scalar >>= 1
        return result

    def __str__(self):
        return f"y² = x³ + ({self.a})x + ({self.b}) over Q"
````

````python
def elliptic_curve_demo():
    """y²=x³−x üzərində dörd elementli altqrupu dəqiq hesablayır."""
    curve = EllipticCurve(-1, 0)
    O, P, Q, R = None, (-1, 0), (0, 0), (1, 0)
    points = [O, P, Q, R]

    assert all(point is None or curve.is_on_curve(*point) for point in points)
    assert curve.add(P, Q) == (Fraction(1), Fraction(0))
    assert curve.add(P, R) == (Fraction(0), Fraction(0))
    assert curve.add(Q, R) == (Fraction(-1), Fraction(0))
    assert all(curve.add(point, point) is None for point in (P, Q, R))

    print("✅ y²=x³−x: {O, P, Q, R} qapalı altqrupdur")
    print("   P+Q=R, P+R=Q, Q+R=P və 2P=2Q=2R=O")


elliptic_curve_demo()
````

````python
def point_addition_demo():
    """Klein dördlü altqrupunun tam toplama cədvəlini qurur."""
    curve = EllipticCurve(-1, 0)
    named_points = {
        "O": None,
        "P": (Fraction(-1), Fraction(0)),
        "Q": (Fraction(0), Fraction(0)),
        "R": (Fraction(1), Fraction(0)),
    }
    point_names = {point: name for name, point in named_points.items()}

    table = []
    for left_name, right_name in product(named_points, repeat=2):
        result = curve.add(named_points[left_name], named_points[right_name])
        assert result in point_names
        table.append((left_name, right_name, point_names[result]))

    for left, right, result in table:
        print(f"{left} + {right} = {result}")
    return table


addition_table = point_addition_demo()
assert len(addition_table) == 16
````

## 🔁 Nöqtəni ikiqat artırma (10 dəq)

<div style="background-color: #f0f8ff; padding: 15px; border-radius: 10px; border-left: 5px solid #4682b4;">
<h4>📐 Nöqtəni ikiqat artırma</h4>
<p>$P$ nöqtəsini ikiqat artırmaq ($2P$) üçün:</p>
<ol>
    <li>$P$-dən keçən toxunan xətt çəkilir</li>
    <li>Bu xətt əyrini ikinci bir nöqtədə kəsir ($R$)</li>
    <li>$2P$, $R$-in $x$-oxuna nəzərən simmetrik nöqtəsidir ($-R$)</li>
</ol>
</div>

````python
def point_doubling_demo():
    """y²=x³+1 üzərində P=(2,3) nöqtəsinin ilk altı qatını göstərir."""
    curve = EllipticCurve(0, 1)
    point = (2, 3)
    multiples = [curve.multiply(point, k) for k in range(1, 7)]
    expected = [(2, 3), (0, 1), (-1, 0), (0, -1), (2, -3), None]
    assert multiples == expected

    for k, multiple in enumerate(multiples, 1):
        print(f"{k}P = {multiple if multiple is not None else 'O'}")
    return multiples


rational_multiples = point_doubling_demo()
````

## ∞ Sonsuzluq nöqtəsi və qrup xassələri (10 dəq)

<div style="background-color: #f0f8ff; padding: 15px; border-radius: 10px; border-left: 5px solid #4682b4;">
<h4>∞ Sonsuzluq nöqtəsi $\mathcal{O}$</h4>
<p>Elliptik əyrilər üzərində nöqtələr çoxluğuna əlavə olaraq <b>sonsuzluq nöqtəsi</b> $\mathcal{O}$ daxil edilir. Bu nöqtə qrupun neytral elementi rolunu oynayır:</p>
<ul>
    <li>$P + \mathcal{O} = P$</li>
    <li>$\mathcal{O} + P = P$</li>
    <li>$\mathcal{O} + \mathcal{O} = \mathcal{O}$</li>
    <li>$P + (-P) = \mathcal{O}$</li>
</ul>
</div>

<div style="background-color: #f0f8ff; padding: 15px; border-radius: 10px; border-left: 5px solid #4682b4; margin-top: 10px;">
<h4>🧩 Qrup xassələri</h4>
<p>Elliptik əyri üzərində nöqtələr çoxluğu (sonsuzluq nöqtəsi daxil olmaqla) Abel qrupu təşkil edir:</p>
<ol>
    <li><b>Qapalılıq:</b> $P + Q$ həmişə əyri üzərindədir</li>
    <li><b>Assosiativlik:</b> $(P + Q) + R = P + (Q + R)$</li>
    <li><b>Neytral element:</b> $P + \mathcal{O} = \mathcal{O} + P = P$</li>
    <li><b>Tərs element:</b> $P + (-P) = \mathcal{O}$</li>
    <li><b>Kommutativlik:</b> $P + Q = Q + P$</li>
</ol>
</div>

````python
def group_properties_demo():
    """{O,P,Q,R} altqrupunun Abel qrupu xassələrini yoxlayır."""
    curve = EllipticCurve(-1, 0)
    points = [None, (-1, 0), (0, 0), (1, 0)]

    assert all(curve.add(a, b) in points for a, b in product(points, repeat=2))
    assert all(curve.add(point, None) == curve.add(None, point) == point for point in points)
    assert all(curve.add(point, curve.negate(point)) is None for point in points)
    assert all(curve.add(a, b) == curve.add(b, a) for a, b in product(points, repeat=2))
    assert all(
        curve.add(curve.add(a, b), c) == curve.add(a, curve.add(b, c))
        for a, b, c in product(points, repeat=3)
    )

    print("✅ {O,P,Q,R} Abel altqrupudur və C₂×C₂ (Klein dördlü qrupu) ilə izomorfdur")
    print("✅ Hər qeyri-neytral elementin sırası 2-dir; altqrupun generatoru yoxdur")
    return points


klein_subgroup = group_properties_demo()
````

### ✍️ Çalışma 4: Qrup xassələri (1 bal)

1. $P+Q+R$ nəyə bərabərdir?
2. $P$-nin tərs elementini və sırasını tapın.
3. $(P+Q)+R$ ilə $P+(Q+R)$-i müqayisə edin.

## 🔐 Diskret loqarifm problemi ECC-də (ECDLP) (10 dəq)

<div style="background-color: #f0f8ff; padding: 15px; border-radius: 10px; border-left: 5px solid #4682b4;">
<h4>🔐 Elliptik Əyri Diskret Loqarifm Problemi (ECDLP)</h4>
<p>Sonlu meydan üzərində seçilmiş, sırası $n$ olan $\langle G\rangle$ altqrupunda $Q=[k]G$ verildikdə $k\bmod n$ skalyarını tapmaq ECDLP adlanır.</p>
<p>Burada $G$ baza nöqtəsidir. Baza nöqtəsinin sırası və kofaktor kimi domen parametrləri təhlükəsiz tətbiq üçün əyrinin özü qədər vacibdir.</p>
<p><b>Tədris sərhədi:</b> aşağıdakı $p=17$ nümunəsi yalnız hesablamanı göstərir və kriptoqrafik cəhətdən təhlükəsiz deyil.</p>
</div>


<div style="background-color: #e8f5e9; padding: 15px; border-radius: 10px; border-left: 5px solid #4caf50; margin-top: 10px;">
<h4>📊 Alqoritmik kontekst</h4>
<ul>
    <li>Ümumi (generic) qrup alqoritmlərindən Pollard rho, $n$ sıralı sadə altqrupda təxminən $O(\sqrt n)$ qrup əməliyyatı tələb edir.</li>
    <li>Bəzi sonlu meydan DLP-ləri üçün xüsusi indeks-hesablama üsulları subeksponensial ola bilər.</li>
    <li>Bu fərq yalnız düzgün seçilmiş domen parametrləri ilə mənalıdır; kiçik və ya yoxlanılmamış əyri avtomatik təhlükəsiz deyil.</li>
</ul>
</div>


````python
class EllipticCurveFp:
    """Sadə p üçün E(F_p): y² = x³ + ax + b; O nöqtəsi None-dır."""

    def __init__(self, p, a, b):
        if not isinstance(p, int) or p <= 3 or any(p % d == 0 for d in range(2, math.isqrt(p) + 1)):
            raise ValueError("p, 3-dən böyük sadə ədəd olmalıdır")
        self.p, self.a, self.b = p, a % p, b % p
        if (4 * self.a**3 + 27 * self.b**2) % p == 0:
            raise ValueError("Əyri F_p üzərində singulyardır")

    def is_on_curve(self, point):
        if point is None:
            return True
        if not isinstance(point, tuple) or len(point) != 2:
            return False
        x, y = point
        return isinstance(x, int) and isinstance(y, int) and (y*y - x*x*x - self.a*x - self.b) % self.p == 0

    def _validate(self, point):
        if not self.is_on_curve(point):
            raise ValueError(f"{point} nöqtəsi E(F_{self.p}) üzərində deyil")
        return None if point is None else (point[0] % self.p, point[1] % self.p)

    def negate(self, point):
        point = self._validate(point)
        return None if point is None else (point[0], (-point[1]) % self.p)

    def add(self, p1, p2):
        p1, p2 = self._validate(p1), self._validate(p2)
        if p1 is None:
            return p2
        if p2 is None:
            return p1

        x1, y1 = p1
        x2, y2 = p2
        if x1 == x2 and (y1 + y2) % self.p == 0:
            return None
        if p1 == p2:
            numerator, denominator = 3 * x1*x1 + self.a, 2 * y1
        else:
            numerator, denominator = y2 - y1, x2 - x1

        slope = numerator * pow(denominator % self.p, -1, self.p) % self.p
        x3 = (slope*slope - x1 - x2) % self.p
        result = (x3, (slope * (x1 - x3) - y1) % self.p)
        assert self.is_on_curve(result)
        return result

    def multiply(self, point, scalar):
        point = self._validate(point)
        if isinstance(scalar, bool) or not isinstance(scalar, int):
            raise TypeError("Skalyar tam ədəd olmalıdır")
        if scalar < 0:
            return self.multiply(self.negate(point), -scalar)
        result, addend = None, point
        while scalar:
            if scalar & 1:
                result = self.add(result, addend)
            addend = self.add(addend, addend)
            scalar >>= 1
        return result

    def points(self):
        return [None] + [
            (x, y) for x in range(self.p) for y in range(self.p)
            if self.is_on_curve((x, y))
        ]

    def point_order(self, point):
        point = self._validate(point)
        if point is None:
            return 1
        current = None
        for order in range(1, len(self.points()) + 1):
            current = self.add(current, point)
            if current is None:
                return order
        raise AssertionError("Nöqtənin sırası tapılmadı")


def ecdlp_demo():
    curve = EllipticCurveFp(17, 2, 2)
    G, secret = (5, 1), 7
    points = curve.points()
    Q = curve.multiply(G, secret)

    assert len(points) == 19
    assert curve.point_order(G) == 19
    recovered = next(k for k in range(19) if curve.multiply(G, k) == Q)
    assert recovered == secret

    print("E: y²=x³+2x+2 over F₁₇")
    print(f"#E(F₁₇)={len(points)}, ord(G)={curve.point_order(G)}")
    print(f"Q=7G={Q}; tam axtarışla bərpa edilən k={recovered}")
    return curve, G, Q


finite_curve, G, Q = ecdlp_demo()
````

### ✍️ Çalışma 5: ECDLP (1 bal)

$E/\mathbb{F}_{17}: y^2=x^3+2x+2$ əyrisi və $G=(5,1)$ baza nöqtəsi verilib.

1. $2G,3G,\ldots,10G$ nöqtələrini hesablayın.
2. $Q=(0,6)$ üçün $Q=[k]G$ bərabərliyini ödəyən $0\le k<19$ qiymətini tapın.
3. $G$-nin sırasını və $E(\mathbb F_{17})$ qrupundakı nöqtələrin sayını hesablayın.
4. Bu oyuncaq nümunənin niyə real kriptoqrafiya üçün təhlükəsiz olmadığını izah edin.


````python
# Çalışma 5 — cavablar
curve = EllipticCurveFp(17, 2, 2)
G = (5, 1)
multiples = {k: curve.multiply(G, k) for k in range(1, 11)}

for k, point in multiples.items():
    print(f"{k:2}G = {point}")

target = (0, 6)
solutions = [k for k in range(19) if curve.multiply(G, k) == target]
assert solutions == [7]
assert curve.point_order(G) == 19
assert len(curve.points()) == 19

print(f"Q={target} üçün k={solutions[0]}")
print("ord(G)=#E(F₁₇)=19; qrup sadə sıralı və G tərəfindən yaradılır")
print("p=17 çox kiçikdir: tam axtarış dərhal nəticə verir, buna görə nümunə təhlükəsiz deyil")
````
