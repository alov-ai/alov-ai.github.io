## 📦 Code-based PQC

Code-based schemes decoding hardness və error-correcting code structure-dan istifadə edir. Eyni ailə bütün key/ciphertext ölçülərini və security assumptions-u paylaşmır.


### 4.1 Toy linear code

Hamming(7,4) yalnız single-bit correction invariantını göstərir; McEliece/Niederreiter KEM deyil.


### 4.2 Classic McEliece

Classic McEliece böyük public keys və uzun tarixə malik code-based KEM candidate idi, lakin NIST 2025 seçimində standardization üçün seçilmədi. Başqa ecosystem statusu ayrıca source tələb edir.


```python
def hamming74_encode(data):
 d1,d2,d3,d4=data;p1=d1^d2^d4;p2=d1^d3^d4;p3=d2^d3^d4
 return [p1,p2,d1,p3,d2,d3,d4]
def hamming74_correct(word):
 w=word[:];s1=w[0]^w[2]^w[4]^w[6];s2=w[1]^w[2]^w[5]^w[6];s3=w[3]^w[4]^w[5]^w[6]
 pos=s1+2*s2+4*s3
 if pos:w[pos-1]^=1
 return [w[2],w[4],w[5],w[6]],pos
hamming_cases=0
for value in range(16):
 data=[(value>>i)&1 for i in (3,2,1,0)];code=hamming74_encode(data)
 for pos in range(8):
  damaged=code[:]
  if pos:damaged[pos-1]^=1
  recovered,syndrome=hamming74_correct(damaged);assert recovered==data and syndrome==pos;hamming_cases+=1
assert hamming_cases==128
print({'cases':128,'scheme':'Hamming(7,4) toy','pqc_claim':False})
```

## 🔗 Hash-based signatures

Security hash assumptions, one-time/few-time signatures, Merkle authentication və hypertree constructions-dan asılıdır. Stateful və stateless operational risks fərqlidir.


### 5.1 Əsas sərhəd

Preimage, second-preimage və collision resistance rolları scheme proof/profile üzrə fərqlənir; “yalnız hash olduğu üçün avtomatik quantum-safe” kifayət deyil.


### 5.2 Lamport one-time signature

256-bit digest üçün hər bitə iki secret/preimage pair lazımdır. Eyni OTS key-i ikinci message üçün istifadə etmək secret material açır; state bunu bloklamalıdır.


```python
def lamport_keygen():
 sk=[(secrets.token_bytes(32),secrets.token_bytes(32)) for _ in range(256)]
 pk=[(hashlib.sha256(a).digest(),hashlib.sha256(b).digest()) for a,b in sk]
 return sk,pk
def digest_bits(message): return [(byte>>shift)&1 for byte in hashlib.sha256(message).digest() for shift in range(7,-1,-1)]
def lamport_sign_once(state,message):
 if state['used']: raise ValueError('Lamport key artıq istifadə edilib')
 state['used']=True;return [state['sk'][i][bit] for i,bit in enumerate(digest_bits(message))]
def lamport_verify(pk,message,signature):
 return len(signature)==256 and all(hashlib.sha256(secret).digest()==pk[i][bit] for i,(secret,bit) in enumerate(zip(signature,digest_bits(message))))
lamport_sk,lamport_pk=lamport_keygen();lamport_state={'sk':lamport_sk,'used':False};lamport_sig=lamport_sign_once(lamport_state,b'AS29')
assert lamport_verify(lamport_pk,b'AS29',lamport_sig) and not lamport_verify(lamport_pk,b'changed',lamport_sig)
reuse_rejected=False
try:lamport_sign_once(lamport_state,b'again')
except ValueError:reuse_rejected=True
assert reuse_rejected
print({'digest_bits':256,'valid':True,'tamper_rejected':True,'reuse_rejected':True})
```

### 6.3 Rainbow

Rainbow practical cryptanalysis nəticələrindən sonra NIST tərəfindən standardization üçün seçilmədi. “Broken” konkret parameter submissions və attacks ilə əlaqələndirilməlidir.


```python
def mq_polynomial(x,y,coeffs,q=17):
 a,b,c,d,e,f=coeffs;return (a*x*x+b*x*y+c*y*y+d*x+e*y+f)%q
coeffs=(1,2,3,4,5,6);mq_values={(x,y):mq_polynomial(x,y,coeffs) for x in range(17) for y in range(17)}
assert len(mq_values)==289 and all(0<=v<17 for v in mq_values.values())
print({'evaluations':289,'signature_scheme':False})
```
